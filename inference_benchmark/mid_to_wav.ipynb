```python
def layer_midis_with_midi2audio(midi_paths, output_wav_path, soundfont_path = '~/home/ubuntu/ugrip/andrew/StreamMUSE/inference_benchmark/FluidR3Mono_GM.sf3'):
    """
    Layers multiple MIDI files using Mido and converts the result to a
    WAV file using the midi2audio library.

    Args:
        midi_paths (list[str]): A list of paths for the MIDI files to layer.
        output_wav_path (str): The path for the final layered WAV file.
        soundfont_path (str): The path to your .sf2 or .sf3 SoundFont file.
    """
    if not midi_paths:
        print("Error: No MIDI file paths were provided.")
        return

    # 1. Combine MIDI files in memory with Mido
    print("Combining MIDI files with Mido...")
    # Load the first file as the base
    combined_midi = mido.MidiFile(midi_paths[0])

    # Append all tracks from the subsequent files
    for path in midi_paths[1:]:
        try:
            mid = mido.MidiFile(path)
            for track in mid.tracks:
                combined_midi.tracks.append(track)
        except Exception as e:
            print(f"Could not process {path}: {e}")
            continue
    
    # 2. Save the combined MIDI to a temporary file
    # Using a 'with' statement ensures the file is created and handled cleanly
    with tempfile.NamedTemporaryFile(suffix=".mid", delete=False) as tmp:
        temp_midi_path = tmp.name
        print(f"Saving combined MIDI to a temporary file: {temp_midi_path}")
        combined_midi.save(temp_midi_path)

    try:
        # 3. Convert the single temporary file to audio
        print("Converting temporary MIDI to WAV with midi2audio...")
        fs = FluidSynth(soundfont_path)
        fs.midi_to_audio(temp_midi_path, output_wav_path)
        print(f"✅ Success! Layered audio saved to: {output_wav_path}")

    finally:
        # 4. Clean up and delete the temporary file
        os.remove(temp_midi_path)
        print(f"Temporary file {temp_midi_path} deleted.")
```
