```python
def mid_to_wav(midi_file, soundfont_path):
    """
    Convert MIDI file to WAV using FluidSynth.
    """
    fs = fluidsynth.Synth()
    fs.start(driver='alsa')
    fs.sfload(soundfont_path)

    output_file = os.path.basename(midi_file).replace('.mid', '.wav')
    fs.midi_to_audio(midi_file, output_file)
    
    fs.delete()
    return output_file
```
