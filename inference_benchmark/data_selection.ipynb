```python
import numpy as np
import glob
import os
import sys
from pathlib import Path
```

```python
def get_dataset_paths(dataset_path_str, num_samples=10, verbose=True, random=True):
    path = Path(dataset_path_str)
    if not path.exists():
        raise FileNotFoundError(f"Dataset path {dataset_path_str} does not exist.")
    if not path.is_dir():
        raise NotADirectoryError(f"Dataset path {dataset_path_str} is not a directory.")

    dataset_paths = sorted(glob.glob('*.mid', root_dir=path))
    if not dataset_paths:
        raise ValueError(f"No MIDI files found in the dataset path {dataset_path_str}.")
    if len(dataset_paths) < num_samples:
        raise ValueError(f"Requested number of samples ({num_samples}) exceeds available files ({len(dataset_paths)}).")
    
    if random:
        selected_paths = np.random.choice(dataset_paths, num_samples, replace=False)
    else:
        selected_paths = dataset_paths[:num_samples]
    absolute_paths = [path / p for p in selected_paths]
    if verbose:
        print(f"Selected {len(absolute_paths)} files from {dataset_path_str}:")
        for p in absolute_paths:
            print(f"{p}")
    return absolute_paths

def copy_to_destination(paths, destination):
    destination_path = Path(destination)
    if not destination_path.exists():
        destination_path.mkdir(parents=True, exist_ok=True)
        print(f"Created destination directory: {destination_path}")

    for p in paths:
        dest_file = destination_path / p.name
        if not dest_file.exists():
            os.system(f'cp "{p}" "{dest_file}"')
        else:
            print(f"File {dest_file} already exists, skipping copy.")
    print(f"Copied {len(paths)} files to {destination_path}.")
```

```python
# Selecting 10 samples (original) from the POP909 dataset
SEED = 42
np.random.seed(SEED)
mel_paths = get_dataset_paths('../../../formatted_dataset/POP909-Dataset/mel', 100, random=True)
acc_paths = [Path(str(p).replace('mel', 'acc')) for p in mel_paths]
copy_to_destination(mel_paths, 'inputs/pop909_dataset/mel')
copy_to_destination(acc_paths, 'inputs/pop909_dataset/acc')
```
