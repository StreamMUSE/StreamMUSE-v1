```python
# Investigate the tempo and timing issue
print("=== INVESTIGATING TEMPO AND TIMING ===")

import music21 as m21
import pretty_midi

midi_file = 'outputs/0722_full_run/real_offline_script/pop909_dataset/024output.mid'

# Check with pretty_midi
pm = pretty_midi.PrettyMIDI(midi_file)
print(f"Pretty_midi total time: {pm.get_end_time():.2f} seconds")
print(f"Pretty_midi downbeats: {pm.get_downbeats()[:10]}")  # First 10 downbeats
print(f"Number of downbeats: {len(pm.get_downbeats())}")

# Check with music21
score = m21.converter.parse(midi_file)
print(f"\nMusic21 score duration: {score.duration.quarterLength} quarter lengths")

# Check tempo markings
tempo_markings = list(score.iter().getElementsByClass('MetronomeMark'))
print(f"Tempo markings found: {len(tempo_markings)}")
for i, tm in enumerate(tempo_markings):
    print(f"  Tempo {i}: {tm.number} BPM at offset {tm.offset}")

# If no tempo, add default and see what happens
if not tempo_markings:
    print("No tempo found, adding default 120 BPM")
    score.insert(0, m21.tempo.MetronomeMark(number=120))

# Check time signature
time_sigs = score.getTimeSignatures()
print(f"Time signatures: {time_sigs}")

# Calculate expected duration for 25 bars at 120 BPM in 4/4
if time_sigs:
    ts = time_sigs[0]
    bar_duration_ql = ts.barDuration.quarterLength
    print(f"Bar duration: {bar_duration_ql} quarter lengths")
    
    # At 120 BPM, 1 quarter note = 0.5 seconds
    expected_duration_25_bars = 25 * bar_duration_ql * 0.5
    print(f"Expected duration for 25 bars at 120 BPM: {expected_duration_25_bars} seconds")

# Now test chordification with tempo
chords_stream = score.chordify()
for mm in score.iter().getElementsByClass('MetronomeMark'):
    chords_stream.insert(mm.offset, mm)

all_chords = list(chords_stream.recurse().getElementsByClass('Chord'))
if all_chords:
    chords_sorted = sorted(all_chords, key=lambda c: c.offset)
    print(f"\nChords timing analysis:")
    print(f"First chord: offset={chords_sorted[0].offset:.2f}, seconds={chords_sorted[0].seconds:.2f}")
    print(f"Last chord: offset={chords_sorted[-1].offset:.2f}, seconds={chords_sorted[-1].seconds:.2f}")
    print(f"Offset range: {chords_sorted[-1].offset - chords_sorted[0].offset:.2f} quarter lengths")
    print(f"Time range: {chords_sorted[-1].seconds - chords_sorted[0].seconds:.2f} seconds")
    
    # Check if the tempo is being applied correctly
    if chords_sorted[-1].offset > 0:
        implied_tempo = (chords_sorted[-1].offset / chords_sorted[-1].seconds) * 60
        print(f"Implied tempo from chord timing: {implied_tempo:.2f} BPM")
```

```python
# The issue is clear: we should use pretty_midi for time calculations, not music21 .seconds
print("=== UNDERSTANDING THE CUTOFF ISSUE ===")

# The musical_statistics function uses pretty_midi to calculate max_time_sec
pm = pretty_midi.PrettyMIDI(midi_file)
downbeats = pm.get_downbeats()
cutoff_bar = 25

if len(downbeats) > cutoff_bar:
    max_time_sec = downbeats[cutoff_bar]
else:
    max_time_sec = pm.get_end_time()

print(f"Cutoff bar: {cutoff_bar}")
print(f"Max time from pretty_midi: {max_time_sec:.2f} seconds")
print(f"This should be around bar {cutoff_bar}, which makes sense")

# The problem is that we're filtering chords by c.seconds < max_time_sec
# But c.seconds is calculated from music21, which has tempo issues
# We should filter by musical position (offset) instead, not by seconds

# Let's see what happens if we filter by offset correctly
score = m21.converter.parse(midi_file)
time_signatures = score.getTimeSignatures()
ts = time_signatures[0] if time_signatures else m21.meter.TimeSignature('4/4')
bar_duration_ql = ts.barDuration.quarterLength
max_offset_ql = cutoff_bar * bar_duration_ql

print(f"Bar duration: {bar_duration_ql} quarter lengths")
print(f"Max offset for {cutoff_bar} bars: {max_offset_ql} quarter lengths")

# Get chords and filter by offset only
chords_stream = score.chordify()
all_chords = list(chords_stream.recurse().getElementsByClass('Chord'))
chords_sorted = sorted(all_chords, key=lambda c: c.offset)

# Filter by offset only (ignore the broken .seconds)
chords_within_bars = [c for c in chords_sorted if c.offset < max_offset_ql]

print(f"Total chords: {len(chords_sorted)}")
print(f"Chords within {cutoff_bar} bars (by offset): {len(chords_within_bars)}")

if chords_within_bars:
    print(f"First chord: offset={chords_within_bars[0].offset:.2f}")
    print(f"Last chord: offset={chords_within_bars[-1].offset:.2f}")
    print(f"Offset range: {chords_within_bars[-1].offset - chords_within_bars[0].offset:.2f} quarter lengths")

# This should give us the correct filtering
```
